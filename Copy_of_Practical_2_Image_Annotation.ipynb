```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = np.ones((500, 500, 3), dtype=np.uint8) * 255
```

```python
cv2.line(img, (50, 50), (200, 50), (255, 0, 0), 3)
```

```python
cv2.putText(img, "Hello OpenCV", (50, 100),
            cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 2)
```

```python
cv2.circle(img, (100, 200), 50, (0, 255, 0), 3)
```

```python
cv2.rectangle(img, (200, 150), (350, 250), (255, 0, 0), 3)
```

```python
cv2.ellipse(img, (250, 350), (100, 50), 0, 0, 360,
            (0, 0, 255), 3)
```

```python
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("annotated_output.jpg", img)

print("Image annotation completed successfully!")
```
